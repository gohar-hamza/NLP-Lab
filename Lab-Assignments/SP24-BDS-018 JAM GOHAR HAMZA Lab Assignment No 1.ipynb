# Task 1: Regex Fundamentals

A mixed sample text is used below. Different regular expressions are applied to extract digits, capitalized words, four-digit sequences, and words of a specific length.


`````python
text = '''
Hamza visited Islamabad in 2026 with 3 friends.
They bought 12 books for Rs. 4500 and attended Data Science Expo 2025.
Python is useful, and Room B17 had 25 students!
'''

print(text)
`````

### 1(a) Extract All Digits

`````python
digits = re.findall(r'\d', text)

print("Digits:")
print(digits)
`````

**Finding:** `\d` matches one digit at a time, so numbers such as `2026` are returned as separate digits.


### 1(b) Extract Words Starting with a Capital Letter

`````python
capital_words = re.findall(r'\b[A-Z][A-Za-z]*\b', text)

print("Capitalized words:")
print(capital_words)
`````

**Finding:** The pattern selects alphabetic words whose first letter is uppercase, such as `Hamza`, `Islamabad`, and `Python`.


### 1(c) Extract Sequences of Exactly 4 Digits

`````python
four_digit_numbers = re.findall(r'(?<!\d)\d{4}(?!\d)', text)

print("Exactly four digits:")
print(four_digit_numbers)
`````

**Finding:** The quantifier `{4}` requires exactly four digits. The surrounding checks prevent matching four digits inside a longer number.


### 1(d) Extract Words Between 3 and 6 Characters Long

`````python
words_3_to_6 = re.findall(r'\b[A-Za-z]{3,6}\b', text)

print("Words with 3 to 6 letters:")
print(words_3_to_6)
`````

**Finding:** `{3,6}` limits the match to alphabetic words containing from 3 to 6 letters.

**Task 1 Conclusion:** Character classes describe what can be matched, while quantifiers control how many characters are required.


# Task 2: Pattern-Based Extraction

The paragraph below contains emails, Pakistani phone numbers, order IDs, dates, and percentages.


`````python
data_text = '''
Customer Ali placed order ORD-45821 on 14/09/2026.
His email is ali.khan27@gmail.com and his phone number is 0301-7654321.
Sara placed order ORD-90317 on 22/09/2026 and can be contacted at
sara_ahmed@outlook.com or 0334-1122334.
Ali received a 45% discount, while Sara received 12.5%.
For support, email help.shop@company.pk.
'''

print(data_text)
`````

### 2(a) Extract Email Addresses

`````python
emails = re.findall(
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    data_text
)

print("Email addresses:")
print(emails)
`````

### 2(b) Extract Phone Numbers

`````python
phone_numbers = re.findall(r'\b03\d{2}-\d{7}\b', data_text)

print("Phone numbers:")
print(phone_numbers)
`````

### 2(c) Extract Dates in DD/MM/YYYY Format

`````python
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', data_text)

print("Dates:")
print(dates)
`````

### 2(d) Extract Percentage Values

`````python
percentages = re.findall(r'\b\d+(?:\.\d+)?%', data_text)

print("Percentages:")
print(percentages)
`````

### Findings and Conclusion

The patterns correctly separate different types of information from the same paragraph.

- Emails are recognized by the `name@domain` structure.
- Phone numbers follow the required `03XX-XXXXXXX` format.
- Dates are selected in `DD/MM/YYYY` format.
- The percentage pattern handles both whole and decimal values.

**Conclusion:** Regex is useful for quickly extracting structured information from unstructured text.


# Task 3: Word Tokenization (NLTK vs spaCy)

The same paragraph is tokenized using both libraries. It contains four sentences and contractions such as **don't**, **it's**, **we'll**, and **you're**.


`````python
paragraph = '''
I don't think it's difficult to learn NLP.
Dr. Sara said, "We'll test tokenization today."
Visit https://example.com if you're curious.
It's useful, isn't it?
'''

print(paragraph)
`````

### 3(a) Tokenization Using NLTK

`````python
from nltk.tokenize import word_tokenize

# preserve_line=True lets word_tokenize work without requiring
# the optional Punkt sentence-tokenizer download.
nltk_tokens = word_tokenize(paragraph, preserve_line=True)

print("NLTK Tokens:")
print(nltk_tokens)
print("\nTotal NLTK tokens:", len(nltk_tokens))
`````

### 3(b) Tokenization Using a Blank spaCy Pipeline

`````python
nlp_blank = spacy.blank("en")
doc = nlp_blank(paragraph)

spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)
print("\nTotal spaCy tokens:", len(spacy_tokens))
`````

### 3(c) Comparison

Two noticeable differences are:

1. In this example, NLTK keeps some sentence-ending periods attached to words such as `NLP.` and `today.`, while spaCy separates the period as its own token.
2. NLTK splits the URL into pieces such as `https`, `:`, and `//example.com`, while spaCy keeps `https://example.com` as one token.

NLTK also changes quotation marks into ```` and `''`, while blank spaCy keeps the quotation mark character.

### 3(d) Token Count

The code above prints the exact token count produced by each tokenizer.

**Conclusion:** Different tokenizers can split the same text differently because they use different tokenization rules.


# Task 4: Token Attributes

The sentence below contains normal words, numbers, an email address, and a currency symbol.


`````python
sample_sentence = (
    "Ali bought 3 books for $50 and emailed store@example.com "
    "about order 2026."
)

doc = nlp_blank(sample_sentence)

print(sample_sentence)
`````

### 4(a–c) Display Token Attributes

For every token:
- `is_alpha` checks whether it contains alphabetic characters only.
- `like_num` checks whether it looks like a number.
- `like_email` checks whether it looks like an email address.


`````python
print(f"{'TOKEN':<22} {'is_alpha':<10} {'like_num':<10} {'like_email':<12}")
print("-" * 58)

for token in doc:
    print(
        f"{token.text:<22} "
        f"{str(token.is_alpha):<10} "
        f"{str(token.like_num):<10} "
        f"{str(token.like_email):<12}"
    )
`````

### 4(d) Identify Currency Tokens

`````python
currency_tokens = [token.text for token in doc if token.is_currency]

print("Currency tokens:")
print(currency_tokens)
`````

### Findings and Conclusion

Words such as `Ali` and `books` have `is_alpha=True`. Numbers such as `3`, `50`, and `2026` have `like_num=True`. The complete email is recognized by `like_email`, and `$` is identified as a currency token.

**Conclusion:** spaCy token attributes provide a simple way to classify tokens without writing a separate regex for every case.


# Task 5: Sentence Segmentation ( Handling Abbreviations)

This example contains abbreviations such as **Dr.**, **e.g.**, and **Mr.** inside sentences.

The code first tries to load the trained `en_core_web_sm` pipeline. A fallback sentencizer is included only so the notebook can still run on a computer where that model is not installed.


`````python
segmentation_text = '''
Dr. Ayesha Khan works at a hospital in Islamabad.
She studies NLP, e.g. text classification and tokenization, for medical notes.
Mr. Ahmed joined her project last week.
They presented the results on Friday.
'''

try:
    nlp = spacy.load("en_core_web_sm")
    pipeline_used = "en_core_web_sm"
except OSError:
    # Fallback for systems where the trained model is not installed
    nlp = spacy.blank("en")
    nlp.add_pipe("sentencizer")
    pipeline_used = "blank English + sentencizer (fallback)"

doc = nlp(segmentation_text)

sentences = [sent.text.strip() for sent in doc.sents]

print("Pipeline used:", pipeline_used)
print("\nDetected sentences:")
for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")

print("\nTotal sentences detected:", len(sentences))
`````

### Observation

The abbreviations **Dr.**, **Mr.**, and **e.g.** did not create incorrect sentence breaks in this example. The paragraph was correctly separated into four sentences.

> To use the exact trained model required by the assignment on a local machine, install it once with:  
> `python -m spacy download en_core_web_sm`

**Task 5 Conclusion:** Sentence segmentation should consider abbreviations so that a period inside an abbreviation is not automatically treated as the end of a sentence.
