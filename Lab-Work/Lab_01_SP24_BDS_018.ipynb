## Task 06: String Cleaning Methods
**Topic:** `lower()`, `upper()`, `strip()`, `split()`, `join()`, `replace()`

### (a) Clean text step by step

```python
text = " MACHINE Learning is AMAZING!!! "

stripped_text = text.strip()
print("After strip():", stripped_text)

lower_text = stripped_text.lower()
print("After lower():", lower_text)

cleaned_text = lower_text.replace("!", "")
print("After replace():", cleaned_text)
```

### (b) Tokenize, uppercase, and join words

```python
tokens = cleaned_text.split()
print("Tokens:", tokens)

uppercase_words = [word.upper() for word in tokens]
print("Uppercase words:", uppercase_words)

joined_text = " | ".join(uppercase_words)
print("Joined text:", joined_text)
```

### (c) Replace a word and find its position

```python
sentence = "I love Java and Java is popular"

updated_sentence = sentence.replace("Java", "Python")
print("Updated sentence:", updated_sentence)

position = updated_sentence.find("Python")
print('Starting position of first "Python":', position)
```

### (d) Create a text normalization function

```python
def normalize(text):
    return text.strip().lower().replace(",", "").replace(".", "")

messy_text_1 = "  Hello, NLP World.  "
messy_text_2 = " Python, Is Great. "

print(normalize(messy_text_1))
print(normalize(messy_text_2))
```

## Task 07: String Validation Methods
**Topic:** `find()`, `startswith()`, `endswith()`, `isdigit()`, `isalpha()`

### (a) Validate the start/end of text and find a word

```python
text = "NLP is the future of AI research"

print('Starts with "NLP":', text.startswith("NLP"))
print('Ends with "research":', text.endswith("research"))
print('Position of "future":', text.find("future"))
```

### (b) Classify tokens as alphabetic, numeric, or mixed

```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(f"{token}: ALPHABETIC")
    elif token.isdigit():
        print(f"{token}: NUMERIC")
    else:
        print(f"{token}: MIXED")
```

### (c) Separate words, numbers, and mixed tokens

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for item in words:
    if item.isalpha():
        pure_words.append(item)
    elif item.isdigit():
        pure_numbers.append(item)
    else:
        mixed_tokens.append(item)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)
```

### (d) Create a token validation function

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    return "MIXED"

test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in test_tokens:
    print(token, "→", validate_token(token))
```

## Task 08: Functions and Lambda for NLP Pipelines
**Topic:** Functions and Lambda — Reusable Text Processing

### (a) Regular function and equivalent lambda

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = " HELLO Natural Language Processing "

print("Regular function:", clean(sample))
print("Lambda function:", clean_lambda(sample))
```

### (b) Tokenize and remove short words

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [token for token in tokens if len(token) > n]

sentence = "I love working with NLP tools in Python daily"

result = remove_short(tokenize(clean(sentence)), 3)
print("Result:", result)
```

### (c) Create a word-length dictionary

```python
def word_lengths(tokens):
    return {word: len(word) for word in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]

length_dictionary = word_lengths(tokens)
print(length_dictionary)
```

### (d) Build a complete NLP pipeline

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

sentence_1 = "Natural Language Processing is very useful"
sentence_2 = "Python makes text analysis simple and powerful"

print("Pipeline output 1:", nlp_pipeline(sentence_1, 3))
print("Pipeline output 2:", nlp_pipeline(sentence_2, 4))
```

## Task 09: `map()`, `filter()`, `sorted()` and `Counter`
**Topic:** `map()`, `filter()`, `sorted()`, Lambda, `collections.Counter`

### (a) Use `map()` with lambda functions

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lowercase_words = list(map(lambda word: word.lower(), words))
word_sizes = list(map(lambda word: len(word), words))

print("Lowercase words:", lowercase_words)
print("Word lengths:", word_sizes)
```

### (b) Use `filter()` with lambda functions

```python
words_long = list(filter(lambda word: len(word) > 3, words))
words_with_t = list(filter(lambda word: "t" in word.lower(), words))

print("Words longer than 3 characters:", words_long)
print('Words containing "t":', words_with_t)
```

### (c) Sort words in three different ways

```python
alphabetical = sorted(words)
length_ascending = sorted(words, key=len)
length_descending = sorted(words, key=len, reverse=True)

print("Alphabetical:", alphabetical)
print("Length ascending:", length_ascending)
print("Length descending:", length_descending)
```

### (d) Count word frequencies using `Counter`

```python
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
word_list = sentence.split()

freq = Counter(word_list)

print("Full Counter:", freq)
print("3 most common words:", freq.most_common(3))
print('Frequency of "python":', freq["python"])
print("Total number of words:", len(word_list))
print("Unique vocabulary size:", len(freq))
```
